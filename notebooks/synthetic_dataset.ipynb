```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
)
```

```python
DATA_PATH = "../data/synthetic-cart-data.csv"
RANDOM_STATE = 42
```

```python
df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
print("Target distribution:\n", df["is_abandoned"].value_counts(normalize=True))
```

```python
drop_cols = ["session_id", "user_id", "is_abandoned"]
X = df.drop(columns=drop_cols)
y = df["is_abandoned"]
 
numeric_features = [
    "cart_value",
    "avg_price_before_cart",
    "previous_purchase_count",
    "user_purchase_rate",
    "days_since_last_purchase",
    "checkout_page_visited",
    "payment_attempted",
    "payment_failed",
    "shipping_cost_shown",
    "coupon_code_attempted",
    "coupon_applied_successfully",
    "exit_intent_detected",
    "time_on_payment_page_seconds",
]
categorical_features = ["device_type", "traffic_source"]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", drop="first"), categorical_features),
    ]
)
```

```python
# ------------------------------------------------------------------
# 3. Train / test split
# ------------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
```

```python
# ------------------------------------------------------------------
# 4. Baseline
# ------------------------------------------------------------------
baseline = DummyClassifier(strategy="most_frequent", random_state=RANDOM_STATE)
baseline.fit(X_train, y_train)
print("\nBaseline accuracy:", accuracy_score(y_test, baseline.predict(X_test)))
 
 
def evaluate(name, pipeline, X_test, y_test):
    pred = pipeline.predict(X_test)
    proba = pipeline.predict_proba(X_test)[:, 1]
    print(f"\n--- {name} ---")
    print("Accuracy :", accuracy_score(y_test, pred))
    print("Precision:", precision_score(y_test, pred))
    print("Recall   :", recall_score(y_test, pred))
    print("F1 Score :", f1_score(y_test, pred))
    print("ROC-AUC  :", roc_auc_score(y_test, proba))
    print("Confusion matrix:\n", confusion_matrix(y_test, pred))
    return pipeline
 
```

```python
# ------------------------------------------------------------------
# 5. Logistic Regression
# ------------------------------------------------------------------
log_reg_pipeline = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        ("model", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)),
    ]
)
log_reg_pipeline.fit(X_train, y_train)
evaluate("Logistic Regression", log_reg_pipeline, X_test, y_test)
```

```python
# ------------------------------------------------------------------
# 6. Random Forest
# ------------------------------------------------------------------
rf_pipeline = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        ("model", RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1)),
    ]
)
rf_pipeline.fit(X_train, y_train)
evaluate("Random Forest", rf_pipeline, X_test, y_test)
```

```python
# ------------------------------------------------------------------
# 7. Gradient Boosting (primary model, matches the main project's choice)
# ------------------------------------------------------------------
gb_pipeline = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        (
            "model",
            GradientBoostingClassifier(
                n_estimators=200, learning_rate=0.05, max_depth=3, random_state=RANDOM_STATE
            ),
        ),
    ]
)
gb_pipeline.fit(X_train, y_train)
evaluate("Gradient Boosting", gb_pipeline, X_test, y_test)
```

```python
# ------------------------------------------------------------------
# 8. Cross-validation for a stable estimate (Gradient Boosting)
# ------------------------------------------------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scores = cross_validate(
    gb_pipeline, X, y, cv=cv, scoring=["accuracy", "precision", "recall", "f1", "roc_auc"]
)
print("\n--- 5-fold cross-validation (Gradient Boosting) ---")
for metric in ["test_accuracy", "test_precision", "test_recall", "test_f1", "test_roc_auc"]:
    print(f"{metric}: mean={scores[metric].mean():.4f} std={scores[metric].std():.4f}")
```

```python
# ------------------------------------------------------------------
# 9. Feature importance (which signals actually drove performance)
# ------------------------------------------------------------------
feature_names = numeric_features + list(
    gb_pipeline.named_steps["preprocess"]
    .named_transformers_["cat"]
    .get_feature_names_out(categorical_features)
)
importances = pd.Series(
    gb_pipeline.named_steps["model"].feature_importances_, index=feature_names
).sort_values(ascending=False)
print("\nFeature importances (Gradient Boosting):")
print(importances)
 
print(
    "\nREMINDER: These results come from a SYNTHETIC dataset with injected "
    "checkout/payment signals. Label this clearly as a hypothetical "
    "simulation in any report -- it does not represent the real "
    "e-commerce clickstream dataset used for the main project."
)
```
