```python
import pandas as pd
import numpy as np


# ============================================================
# 1. LOAD CLEAN EVENTS
# ============================================================

events = pd.read_csv(
    "../data/clean_events.csv"
)

events["event_time"] = pd.to_datetime(
    events["event_time"],
    utc=True
)

events = events.sort_values(
    ["user_id", "event_time"]
).reset_index(drop=True)


# ============================================================
# 2. RECONSTRUCT 30-MINUTE SESSIONS
# ============================================================

events["previous_event_time"] = (
    events
    .groupby("user_id")["event_time"]
    .shift(1)
)

events["time_gap_minutes"] = (
    (
        events["event_time"]
        - events["previous_event_time"]
    )
    .dt.total_seconds()
    / 60
)

events["new_session"] = (
    events["time_gap_minutes"].isna()
    |
    (events["time_gap_minutes"] > 30)
)

events["session_number"] = (
    events
    .groupby("user_id")["new_session"]
    .cumsum()
)

events["reconstructed_session_id"] = (
    events["user_id"].astype(str)
    + "_"
    + events["session_number"].astype(str)
)


# ============================================================
# 3. LOAD V1 MODEL DATASET
# ============================================================

df = pd.read_csv(
    "../data/model_ready_cart_events_v1.csv"
)

df["cart_time"] = pd.to_datetime(
    df["cart_time"],
    utc=True
)

df = df.reset_index(drop=True)

df["cart_row_id"] = np.arange(
    len(df)
)

print(
    "Original model dataset:",
    df.shape
)


# ============================================================
# 4. CART INFORMATION
# ============================================================

cart_base = df[
    [
        "cart_row_id",
        "reconstructed_session_id",
        "product_id",
        "category_id",
        "brand",
        "cart_time"
    ]
].copy()

cart_base = cart_base.rename(
    columns={
        "product_id": "cart_product_id",
        "category_id": "cart_category_id",
        "brand": "cart_brand"
    }
)


# ============================================================
# 5. EVENTS NEEDED FOR POST-CART ANALYSIS
# ============================================================

event_data = events[
    [
        "reconstructed_session_id",
        "event_time",
        "event_type",
        "product_id",
        "category_id",
        "brand"
    ]
].copy()


# ============================================================
# 6. MERGE CART WITH EVENTS FROM SAME SESSION
# ============================================================

post = cart_base.merge(
    event_data,
    on="reconstructed_session_id",
    how="left"
)


# ============================================================
# 7. CALCULATE TIME AFTER CART
# ============================================================

post["minutes_after_cart"] = (
    (
        post["event_time"]
        - post["cart_time"]
    )
    .dt.total_seconds()
    / 60
)


# Only events AFTER cart
# and within 30 minutes

post = post[
    (post["minutes_after_cart"] > 0)
    &
    (post["minutes_after_cart"] <= 30)
].copy()


# ============================================================
# 8. REMOVE PURCHASE EVENTS
# ============================================================
# Very important:
# purchase itself was NOT used as a feature.

post = post[
    post["event_type"] != "purchase"
].copy()


print(
    "Post-cart non-purchase events:",
    len(post)
)


# ============================================================
# 9. BASIC EVENT FLAGS
# ============================================================

post["is_view"] = (
    post["event_type"] == "view"
).astype(int)

post["is_cart"] = (
    post["event_type"] == "cart"
).astype(int)


# ============================================================
# 10. SAME PRODUCT BEHAVIOUR
# ============================================================

post["same_product_view"] = (
    (post["event_type"] == "view")
    &
    (
        post["product_id"]
        ==
        post["cart_product_id"]
    )
).astype(int)


post["same_product_cart"] = (
    (post["event_type"] == "cart")
    &
    (
        post["product_id"]
        ==
        post["cart_product_id"]
    )
).astype(int)


# ============================================================
# 11. DIFFERENT PRODUCT VIEW
# ============================================================

post["different_product_view"] = (
    (post["event_type"] == "view")
    &
    (
        post["product_id"]
        !=
        post["cart_product_id"]
    )
).astype(int)


# ============================================================
# 12. CATEGORY SWITCH
# ============================================================

post["category_switch"] = (
    post["category_id"]
    !=
    post["cart_category_id"]
).astype(int)


# ============================================================
# 13. BRAND SWITCH
# ============================================================

post["brand_switch"] = (
    post["brand"].astype(str)
    !=
    post["cart_brand"].astype(str)
).astype(int)


# ============================================================
# 14. AGGREGATE POST-CART FEATURES
# ============================================================

post_features = (
    post
    .groupby("cart_row_id")
    .agg(

        post_event_count_30m=(
            "event_type",
            "size"
        ),

        post_view_count_30m=(
            "is_view",
            "sum"
        ),

        post_cart_count_30m=(
            "is_cart",
            "sum"
        ),

        same_product_views_after_cart_30m=(
            "same_product_view",
            "sum"
        ),

        same_product_carts_after_cart_30m=(
            "same_product_cart",
            "sum"
        ),

        different_product_views_after_cart_30m=(
            "different_product_view",
            "sum"
        ),

        unique_products_after_cart_30m=(
            "product_id",
            "nunique"
        ),

        unique_categories_after_cart_30m=(
            "category_id",
            "nunique"
        ),

        unique_brands_after_cart_30m=(
            "brand",
            "nunique"
        ),

        category_switch_after_cart_30m=(
            "category_switch",
            "max"
        ),

        brand_switch_after_cart_30m=(
            "brand_switch",
            "max"
        ),

        first_post_event_minutes=(
            "minutes_after_cart",
            "min"
        ),

        last_post_event_minutes=(
            "minutes_after_cart",
            "max"
        )

    )
    .reset_index()
)


# ============================================================
# 15. MERGE FEATURES WITH V1
# ============================================================

model_v4 = df.merge(
    post_features,
    on="cart_row_id",
    how="left"
)


# ============================================================
# 16. FILL NO POST-CART ACTIVITY WITH 0
# ============================================================

count_features = [
    "post_event_count_30m",
    "post_view_count_30m",
    "post_cart_count_30m",
    "same_product_views_after_cart_30m",
    "same_product_carts_after_cart_30m",
    "different_product_views_after_cart_30m",
    "unique_products_after_cart_30m",
    "unique_categories_after_cart_30m",
    "unique_brands_after_cart_30m",
    "category_switch_after_cart_30m",
    "brand_switch_after_cart_30m"
]

model_v4[count_features] = (
    model_v4[count_features]
    .fillna(0)
)


# ============================================================
# 17. NO-ACTIVITY FLAG
# ============================================================

model_v4["no_activity_after_cart_30m"] = (
    model_v4["post_event_count_30m"] == 0
).astype(int)


# ============================================================
# 18. FIRST / LAST EVENT TIMING
# ============================================================

model_v4[
    "first_post_event_minutes"
] = (
    model_v4[
        "first_post_event_minutes"
    ]
    .fillna(-1)
)

model_v4[
    "last_post_event_minutes"
] = (
    model_v4[
        "last_post_event_minutes"
    ]
    .fillna(-1)
)


# ============================================================
# 19. ACTIVITY SPAN
# ============================================================

model_v4[
    "post_activity_span_minutes"
] = np.where(

    model_v4[
        "post_event_count_30m"
    ] > 0,

    model_v4[
        "last_post_event_minutes"
    ]
    -
    model_v4[
        "first_post_event_minutes"
    ],

    0
)


# ============================================================
# 20. POST-CART / VIEW RATIO
# ============================================================

model_v4[
    "post_cart_to_view_ratio"
] = (
    model_v4[
        "post_cart_count_30m"
    ]
    /
    (
        model_v4[
            "post_view_count_30m"
        ] + 1
    )
)


# ============================================================
# 21. REMOVE TEMPORARY ID
# ============================================================

model_v4 = model_v4.drop(
    columns=[
        "cart_row_id"
    ]
)


# ============================================================
# 22. VALIDATE
# ============================================================

print("\n" + "=" * 60)

print(
    "V4 shape:",
    model_v4.shape
)

print(
    "Missing values:",
    model_v4.isnull()
    .sum()
    .sum()
)

print(
    "\nTarget:"
)

print(
    model_v4[
        "is_abandoned"
    ]
    .value_counts()
)


# ============================================================
# 23. SAVE V4
# ============================================================

model_v4.to_csv(
    "../data/new_model_ready_cart_events_v4.csv",
    index=False
)

print(
    "\nSaved successfully:"
)

print(
    "../data/new_model_ready_cart_events_v4.csv"
)
```

```python
import pandas as pd

from catboost import CatBoostClassifier

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)


# ============================================================
# LOAD V4
# ============================================================

df = pd.read_csv(
    "../data/new_model_ready_cart_events_v4.csv"
)

print("Dataset:", df.shape)


# ============================================================
# TARGET
# ============================================================

y = df["is_abandoned"].copy()


# ============================================================
# FEATURES
# Keep PRODUCT + CATEGORY IDs
# ============================================================

X = df.drop(
    columns=[
        "is_abandoned",
        "reconstructed_session_id",
        "user_id",
        "cart_time"
    ]
).copy()


# ============================================================
# CATEGORICAL FEATURES
# ============================================================

categorical_features = [
    "product_id",
    "category_id",
    "category_code",
    "brand"
]

for col in categorical_features:
    X[col] = (
        X[col]
        .fillna("unknown")
        .astype(str)
    )


print(
    "Total features:",
    X.shape[1]
)

print(
    "Categorical:",
    categorical_features
)


# ============================================================
# TRAIN / TEST
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# CATBOOST
# ============================================================

model = CatBoostClassifier(

    iterations=1200,

    depth=8,

    learning_rate=0.05,

    loss_function="Logloss",

    eval_metric="AUC",

    random_seed=42,

    l2_leaf_reg=6,

    random_strength=1,

    allow_writing_files=False
)


model.fit(

    X_train,
    y_train,

    cat_features=categorical_features,

    eval_set=(
        X_test,
        y_test
    ),

    early_stopping_rounds=100,

    verbose=100
)


# ============================================================
# PREDICTION
# ============================================================

pred = (
    model
    .predict(X_test)
    .astype(int)
    .flatten()
)

prob = (
    model
    .predict_proba(X_test)[:, 1]
)


# ============================================================
# METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    pred
)

precision = precision_score(
    y_test,
    pred
)

recall = recall_score(
    y_test,
    pred
)

f1 = f1_score(
    y_test,
    pred
)

auc = roc_auc_score(
    y_test,
    prob
)


print("\n" + "=" * 60)
print("CATBOOST + PRODUCT/CATEGORY ID")
print("=" * 60)

print(
    f"Accuracy  : {accuracy*100:.2f}%"
)

print(
    f"Precision : {precision*100:.2f}%"
)

print(
    f"Recall    : {recall*100:.2f}%"
)

print(
    f"F1 Score  : {f1*100:.2f}%"
)

print(
    f"ROC-AUC   : {auc*100:.2f}%"
)


print("\nConfusion Matrix:")

print(
    confusion_matrix(
        y_test,
        pred
    )
)


# ============================================================
# FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({
    "feature": X.columns,
    "importance": model.get_feature_importance()
})

importance = (
    importance
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nTop 25 Features:")

print(
    importance.head(25)
)
```
